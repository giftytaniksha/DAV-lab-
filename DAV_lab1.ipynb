```python
# ============================================================
# LAB ASSIGNMENT 1
# DATA ANALYTICS AND VISUALIZATION
# Significant Earthquakes (1995-2023)
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns when printing DataFrames
pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# Upload earthquake_1995-2023.csv to Google Colab before running
df = pd.read_csv("earthquake_1995-2023.csv")

print("Dataset loaded successfully.")
print("Original dataset shape:", df.shape)


# ============================================================
# 3. DATA PROFILING
# ============================================================

print("\n========== FIRST FIVE RECORDS ==========")
display(df.head())

print("\n========== DATASET SHAPE ==========")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA INFORMATION ==========")
df.info()


# ============================================================
# 4. DUPLICATE DETECTION
# ============================================================

# ------------------------------------------------------------
# 4.1 Fully identical rows
# ------------------------------------------------------------

print("\n========== FULLY IDENTICAL DUPLICATES ==========")

full_duplicates = df.duplicated().sum()

print("Number of fully identical duplicate rows:", full_duplicates)


# Remove fully identical duplicates
rows_before = len(df)

df = df.drop_duplicates().copy()

rows_after = len(df)

print("Rows before removing fully identical duplicates:", rows_before)
print("Rows after removing fully identical duplicates:", rows_after)
print("Rows removed:", rows_before - rows_after)


# ------------------------------------------------------------
# 4.2 Duplicate earthquakes using title + date_time
# ------------------------------------------------------------

print("\n========== DUPLICATE EARTHQUAKES ==========")

duplicate_key_mask = df.duplicated(
    subset=["title", "date_time"],
    keep=False
)

print(
    "Rows involved in repeated title + date_time combinations:",
    duplicate_key_mask.sum()
)

print("\nDuplicate earthquake records:")

display(
    df.loc[
        duplicate_key_mask,
        ["title", "date_time"]
    ].sort_values(["title", "date_time"])
)


# Count duplicates that will actually be removed
duplicate_earthquakes = df.duplicated(
    subset=["title", "date_time"]
).sum()

print(
    "\nNumber of duplicate earthquake records to remove:",
    duplicate_earthquakes
)


# Remove duplicate earthquakes
rows_before = len(df)

df = df.drop_duplicates(
    subset=["title", "date_time"],
    keep="first"
).copy()

rows_after = len(df)

print("\nRows before duplicate removal:", rows_before)
print("Rows after duplicate removal:", rows_after)
print("Total rows removed:", rows_before - rows_after)


# ============================================================
# 5. MISSING VALUE REPORT
# ============================================================

print("\n========== MISSING VALUE REPORT ==========")

missing_report = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": df.isna().mean() * 100
})

missing_report["Missing Percentage"] = (
    missing_report["Missing Percentage"].round(2)
)

missing_report = missing_report.sort_values(
    by="Missing Count",
    ascending=False
)

display(missing_report)


# ============================================================
# 6. INVESTIGATE PLACEHOLDER ZEROS
# ============================================================

print("\n========== PLACEHOLDER ZERO ANALYSIS ==========")

# According to the assignment, zero can represent
# "not measured" for these numeric columns.
placeholder_zero_columns = [
    "cdi",
    "nst",
    "dmin",
    "gap"
]

for column in placeholder_zero_columns:
    print(
        f"{column}:",
        (df[column] == 0).sum(),
        "zero values"
    )


# tsunami = 0 is a meaningful binary value,
# so we DO NOT convert tsunami=0 to NaN.

print(
    "\ntsunami:",
    (df["tsunami"] == 0).sum(),
    "zero values"
)

print(
    "tsunami=0 is retained because it represents a meaningful "
    "binary value rather than missing data."
)


# ------------------------------------------------------------
# Convert placeholder zeros to NaN
# ------------------------------------------------------------

for column in placeholder_zero_columns:
    df[column] = df[column].replace(0, np.nan)

print("\nPlaceholder zeros converted to NaN.")

print("\nMissing values after zero replacement:")

display(
    df[placeholder_zero_columns].isna().sum()
)


# ============================================================
# 7. MEAN VS MEDIAN ANALYSIS
# ============================================================

# The assignment requires:
# - histogram
# - skewness
# - decision between mean and median
#
# We examine four numeric columns.


# ------------------------------------------------------------
# 7.1 CDI
# ------------------------------------------------------------

print("\n========== CDI ANALYSIS ==========")

cdi_skewness = df["cdi"].skew()
cdi_mean = df["cdi"].mean()
cdi_median = df["cdi"].median()

print("CDI skewness:", round(cdi_skewness, 3))
print("CDI mean:", round(cdi_mean, 3))
print("CDI median:", round(cdi_median, 3))

plt.figure(figsize=(8, 5))

plt.hist(
    df["cdi"].dropna(),
    bins=12,
    edgecolor="black"
)

plt.xlabel("CDI")
plt.ylabel("Frequency")
plt.title("Distribution of CDI")

plt.show()


# ------------------------------------------------------------
# 7.2 NST
# ------------------------------------------------------------

print("\n========== NST ANALYSIS ==========")

nst_skewness = df["nst"].skew()
nst_mean = df["nst"].mean()
nst_median = df["nst"].median()

print("NST skewness:", round(nst_skewness, 3))
print("NST mean:", round(nst_mean, 3))
print("NST median:", round(nst_median, 3))

plt.figure(figsize=(8, 5))

plt.hist(
    df["nst"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.xlabel("NST")
plt.ylabel("Frequency")
plt.title("Distribution of NST")

plt.show()


# ------------------------------------------------------------
# 7.3 DMIN
# ------------------------------------------------------------

print("\n========== DMIN ANALYSIS ==========")

dmin_skewness = df["dmin"].skew()
dmin_mean = df["dmin"].mean()
dmin_median = df["dmin"].median()

print("DMIN skewness:", round(dmin_skewness, 3))
print("DMIN mean:", round(dmin_mean, 3))
print("DMIN median:", round(dmin_median, 3))

plt.figure(figsize=(8, 5))

plt.hist(
    df["dmin"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.xlabel("DMIN")
plt.ylabel("Frequency")
plt.title("Distribution of DMIN")

plt.show()


# ------------------------------------------------------------
# 7.4 GAP
# ------------------------------------------------------------

print("\n========== GAP ANALYSIS ==========")

gap_skewness = df["gap"].skew()
gap_mean = df["gap"].mean()
gap_median = df["gap"].median()

print("GAP skewness:", round(gap_skewness, 3))
print("GAP mean:", round(gap_mean, 3))
print("GAP median:", round(gap_median, 3))

plt.figure(figsize=(8, 5))

plt.hist(
    df["gap"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.xlabel("GAP")
plt.ylabel("Frequency")
plt.title("Distribution of GAP")

plt.show()


# ============================================================
# 8. IMPUTATION DECISIONS
# ============================================================

print("\n========== IMPUTATION DECISIONS ==========")

print("""
CDI:
The skewness is relatively small, so the distribution is reasonably
close to symmetric. Therefore, mean imputation is appropriate.

NST:
The skewness is close to zero, indicating an approximately symmetric
distribution. Therefore, mean imputation is appropriate.

DMIN:
The distribution is strongly right-skewed. The median is less affected
by extreme values and outliers, so median imputation is appropriate.

GAP:
The distribution is highly right-skewed. Therefore, median imputation
is preferred because it is more robust to extreme values.
""")


# ============================================================
# 9. SAVE CDI BEFORE IMPUTATION FOR COMPARISON
# ============================================================

cdi_before_imputation = df["cdi"].copy()


# ============================================================
# 10. NUMERIC IMPUTATION
# ============================================================

# Mean imputation for relatively symmetric distributions
df["cdi"] = df["cdi"].fillna(df["cdi"].mean())

df["nst"] = df["nst"].fillna(df["nst"].mean())


# Median imputation for skewed distributions
df["dmin"] = df["dmin"].fillna(df["dmin"].median())

df["gap"] = df["gap"].fillna(df["gap"].median())


print("\n========== NUMERIC IMPUTATION COMPLETE ==========")

print("Remaining missing values in numeric columns:")

display(
    df[
        ["cdi", "nst", "dmin", "gap"]
    ].isna().sum()
)


# ============================================================
# 11. CATEGORICAL / TEXT MISSING VALUES
# ============================================================

print("\n========== CATEGORICAL MISSING VALUES ==========")

# alert -> mode
alert_mode = df["alert"].mode()[0]

df["alert"] = df["alert"].fillna(alert_mode)

# Text/geographical fields -> Unknown
df["location"] = df["location"].fillna("Unknown")

df["continent"] = df["continent"].fillna("Unknown")

df["country"] = df["country"].fillna("Unknown")


print("Alert missing values filled using mode:", alert_mode)

print("\nRemaining categorical/text missing values:")

display(
    df[
        ["alert", "location", "continent", "country"]
    ].isna().sum()
)


# ============================================================
# 12. FINAL MISSING VALUE REPORT
# ============================================================

print("\n========== FINAL MISSING VALUE REPORT ==========")

final_missing = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": df.isna().mean() * 100
})

final_missing["Missing Percentage"] = (
    final_missing["Missing Percentage"].round(2)
)

display(final_missing)


# ============================================================
# 13. VISUALIZATION 1
# DISTRIBUTION OF EARTHQUAKE MAGNITUDES
# ============================================================

print("\n========== VISUALIZATION 1 ==========")

plt.figure(figsize=(8, 5))

plt.hist(
    df["magnitude"],
    bins=20,
    edgecolor="black"
)

plt.xlabel("Magnitude")
plt.ylabel("Number of Earthquakes")
plt.title("Distribution of Earthquake Magnitudes")

plt.show()


# ============================================================
# 14. VISUALIZATION 2
# EARTHQUAKES BY ALERT LEVEL
# ============================================================

print("\n========== VISUALIZATION 2 ==========")

plt.figure(figsize=(8, 5))

df["alert"].value_counts().plot(
    kind="bar",
    edgecolor="black"
)

plt.xlabel("Alert Level")
plt.ylabel("Number of Earthquakes")
plt.title("Earthquakes by Alert Level")

plt.xticks(rotation=0)

plt.show()


# ============================================================
# 15. VISUALIZATION 3
# CDI BEFORE VS AFTER IMPUTATION
# ============================================================

print("\n========== VISUALIZATION 3 ==========")

plt.figure(figsize=(8, 5))

plt.hist(
    cdi_before_imputation.dropna(),
    bins=12,
    alpha=0.6,
    label="Before Imputation",
    edgecolor="black"
)

plt.hist(
    df["cdi"],
    bins=12,
    alpha=0.6,
    label="After Imputation",
    edgecolor="black"
)

plt.xlabel("CDI")
plt.ylabel("Frequency")
plt.title("CDI Before vs After Mean Imputation")

plt.legend()

plt.show()


# ============================================================
# 16. FINAL VALIDATION
# ============================================================

print("\n========== FINAL VALIDATION ==========")

print("Missing values in every column:")

display(df.isna().sum())


total_missing = df.isna().sum().sum()

print("\nTotal remaining missing values:", total_missing)

print("\nFinal dataset shape:", df.shape)


# ============================================================
# 17. FINAL CHECK
# ============================================================

if total_missing == 0:
    print("\nSUCCESS: The cleaned dataset contains NO missing values.")
else:
    print("\nWARNING: Missing values are still present.")


# ============================================================
# 18. FINAL CLEANED DATASET PREVIEW
# ============================================================

print("\n========== CLEANED DATASET ==========")

display(df.head())


# ============================================================
# 19. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "earthquake_1995-2023_cleaned.csv",
    index=False
)

print(
    "\nCleaned dataset saved as "
    "'earthquake_1995-2023_cleaned.csv'"
)


# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print("\n============================================================")
print("FINAL SUMMARY")
print("============================================================")

print("Original dataset rows      :", 1000)
print("Original dataset columns   :", 19)
print("Fully identical duplicates :", full_duplicates)
print("Duplicate earthquakes      :", duplicate_earthquakes)
print("Final dataset rows         :", df.shape[0])
print("Final dataset columns      :", df.shape[1])
print("Remaining missing values   :", total_missing)

print("============================================================")

if df.shape == (998, 19) and total_missing == 0:
    print("LAB ASSIGNMENT CLEANING COMPLETED SUCCESSFULLY.")
else:
    print("Please review the results above.")
```
